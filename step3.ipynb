```python
station_info = pd.read_csv('sub_station_name.CSV', encoding='cp949')
```

```python
station_info.head()
```

```python
# 자치구 정보에서 각 역명을 개별 행으로 분리
station_info_expanded = station_info.copy()
```

```python
# '해당역(호선)' 컬럼을 ','로 나눈 후 각 역에서 괄호 제거하고 strip
station_info_expanded = station_info_expanded.drop('역개수', axis=1)
station_info_expanded = station_info_expanded.assign(
    역명목록=station_info_expanded['해당역(호선)'].str.split(', ')
).explode('역명목록')
```

```python
# 괄호 제거하고 역명만 추출
station_info_expanded['역명'] = station_info_expanded['역명목록'].str.extract(r'([가-힣0-9]+)\(')
station_info_expanded = station_info_expanded[['자치구', '역명']]
```

```python
# 중복 제거
station_info_expanded = station_info_expanded.drop_duplicates()
```

```python
# subway에 자치구 매핑
subway_with_gu = pd.merge(subway, station_info_expanded, how='left', on='역명')
```
