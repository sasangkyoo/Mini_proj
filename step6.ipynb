```python
wed = wed.drop(columns=['지점', '지점명'])
```

```python
# 서울시 25개 자치구 리스트
valid_gu_list = [
    '강남구', '강동구', '강북구', '강서구', '관악구', '광진구', '구로구', '금천구',
    '노원구', '도봉구', '동대문구', '동작구', '마포구', '서대문구', '서초구',
    '성동구', '성북구', '송파구', '양천구', '영등포구', '용산구', '은평구',
    '종로구', '중구', '중랑구'
]

# 특정 컬럼에서 자치구가 아닌 항목 제거 (예: '자치구' 컬럼이라고 가정)
pol = pol[pol['측정소명'].isin(valid_gu_list)]
```

```python
wed['일자'] = pd.to_datetime(wed['일시'])
pol['일자'] = pd.to_datetime(pol['측정일시'].astype(str), format='%Y%m%d')
```

```python
wed = wed.drop(columns=['일시'])
pol = pol.drop(columns=['측정일시'])
```

```python
pol.head()
```

```python
wed.head()
```

```python
# '측정소명'과 '일자' 기준으로, wed의 전체 날씨 데이터 컬럼과 left join
forcast = wed.merge(pol, how='left', left_on=['자치구', '일자'], right_on=['측정소명', '일자'])
```

```python
# 자치구가 NaN인 경우 측정소명으로 채움
forcast['자치구'] = forcast['자치구'].fillna(forcast['측정소명'])

# 측정소명 컬럼 제거
forcast = forcast.drop(columns=['측정소명'])
```

```python
forcast = forcast[[
    '일자', '자치구',
    '평균기온(°C)', '최저기온(°C)', '최고기온(°C)', '일강수량(mm)',
    '이산화질소농도(ppm)', '오존농도(ppm)', '일산화탄소농도(ppm)',
    '아황산가스농도(ppm)', '미세먼지농도(㎍/㎥)', '초미세먼지농도(㎍/㎥)'
]]
```
